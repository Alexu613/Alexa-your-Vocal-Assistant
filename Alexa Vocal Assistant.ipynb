### Speech-to-Text Transcription with Whisper

Speech-to-Text Transcription with Whisper refers to converting spoken language into written text using Whisper, an automatic speech recognition (ASR) system developed by OpenAI.
Key points:

    - Whisper is an open-source model.

    - It can transcribe audio in multiple languages.

    - It can also do translation from other languages to English.

It’s trained on a large amount of diverse audio data, making it good at understanding real-world accents, background noise, and varied speaking styles.

When we use "rb" in open(), it means:

    "r" = read → you're opening the file to read it

    "b" = binary → you're reading it in binary mode (not text)

    "rb" = read binary

This is used for files like audio, video, or images, where the content isn't plain text — it's raw data (1s and 0s).

So Python reads the file exactly as it is, without trying to turn it into readable text.

## Alexa : Vocal Assistant

🎙️ New project in progress: building an intelligent voice assistant inspired by Alexa, combining OpenAI’s GPT-4 API with Python text-to-speech synthesis. The goal? To enable smooth and natural interaction between humans and machines. A great opportunity to explore the real-time potential of conversational AI! 🤖💬

```python
import openai
import pyttsx3
import os

try:
    #Start a conversation with Alexa
    print('Welcome to Alexa Vocal Assistant !')

    while True:
        #Ask your question here:
        user_prompt = input("What do you want to ask Alexa ?")
        print(f"\n 🙎‍♂️ User : {user_prompt}")

        #Intelligent Assistant Agent's Behaviour:
        behaviour = "You are an intelligent system called Alexa capable of creating a conversation between \
        you and a human and answer any time of question."
        messages = [
            {"role":"system", "content": behaviour}
        ]
        #Add user's question to messages list:
        messages.append(
            {"role":"user", "content": user_prompt}
        )
        
        #Condition to exit/leave the conversation with the agent:
        if user_prompt == "exit":
            print("Thanks for choosing Alexa, Have a good day !")
            break 

        #Create a client to interact with:
        client = openai.OpenAI(api_key=api_key) 
        response = client.chat.completions.create(
            model="gpt-4", #GPT4 Model
            messages = messages,
            max_tokens=150, #Maximum Tokens do not have to exceed 150
            top_p=0.9,
            temperature=0.2
        )
        print(f"\n 🤖 Alexa : {response.choices[0].message.content}")
        print("-------------------------------------------------")
        
        #Save the agent's response:
        agent_response = response.choices[0].message.content
        # Initialization:
        engine = pyttsx3.init()
        # Choose an optional voice:
        voices = engine.getProperty('voices')
        # Let's try for an english voice:
        for voice in voices:
            if "Karen" in voice.name:
                engine.setProperty('voice', voice.id)
                break
            
        # Ask Alexa to say this text out loud !
        engine.say(agent_response)
        engine.runAndWait()

except Exception as e:
    print('Error:', e)
```
